## 步骤 7.4(2):外参初值

**测量法**(粗略,可先用来渲染看看):用 §4.2 读出相机镜头附近一点的基座系坐标,卷尺量高度,手机水平仪量俯角。
**手眼标定法**(更准):前视用 `cv2.calibrateRobotWorldHandEye`,腕部用 `cv2.calibrateHandEye`,走 15–20 个姿态。这需要另外写采集脚本,见文档。

下面用测量法。填你的前视相机在**基座系**下的位置(米)、俯角和偏航(面朝机器人 = 0°)。默认值是作者的标称位姿。

````python
FLIP = np.diag([1, -1, -1])                        # OpenCV 相机系 → USD 相机系
BASE_TO_WORLD = np.array([-0.30, 0.0, 0.415])      # 世界坐标 = 基座坐标 + 这个

def look_pose(pos, pitch_deg, yaw_deg=0.0):
    """OpenCV 约定(x 右、y 下、z 前)下,相机在基座系的 4x4 位姿。
    yaw=0 面朝 -x(面朝机器人),pitch 为下俯角,无侧倾。"""
    th, ps = np.radians(pitch_deg), np.radians(yaw_deg)
    z = np.array([-np.cos(th)*np.cos(ps), -np.cos(th)*np.sin(ps), -np.sin(th)])
    x = np.cross(z, [0, 0, 1]); x /= np.linalg.norm(x)
    y = np.cross(z, x)
    T = np.eye(4); T[:3, :3] = np.c_[x, y, z]; T[:3, 3] = pos
    return T

# ====== 改这里 ======
FRONT_POS_BASE = [1.13, 0, 0.415]   # 基座系 (x, y, z) 米
FRONT_PITCH    = 35.5                    # 下俯角,度
FRONT_YAW      = 0.0                     # 偏航,度
# ====================
T_base_front = look_pose(FRONT_POS_BASE, FRONT_PITCH, FRONT_YAW)
````

## 步骤 7.4(3):换算成渲染器格式

渲染器用 Isaac/USD 相机约定:视线沿 −Z,上方向 +Y,内旋 "XYZ" 欧拉角。
先用标称值做**自检**:应得到 `pos=(0.82, -0.065, 0.8)`、`euler=[0, 68.5, 90]`。

````python
def front_to_rerender(T_base_cam):
    pos = T_base_cam[:3, 3] + BASE_TO_WORLD
    eul = R.from_matrix(T_base_cam[:3, :3] @ FLIP).as_euler("XYZ", degrees=True)
    return pos, eul

def wrist_to_rerender(T_flange_cam):
    # panda_hand = 法兰绕 z 转 -45°
    T_hand_flange = np.eye(4); T_hand_flange[:3, :3] = R.from_euler("z", 45, degrees=True).as_matrix()
    T = T_hand_flange @ T_flange_cam
    return T[:3, 3], R.from_matrix(T[:3, :3] @ FLIP).as_euler("XYZ", degrees=True)

# 自检:标称前视
p, e = front_to_rerender(look_pose([1.12, -0.065, 0.385], 21.5))
print("nominal  pos =", np.round(p, 4), " euler =", np.round(e, 2), "  (期望 [0.82 -0.065 0.8], [0 68.5 90])")

# 你的前视
pos, eul = front_to_rerender(T_base_front)
print("\n你的前视 → isaac_lab_rerender.py 里:")
print(f"  cam_pos = np.array({np.round(pos, 4).tolist()})")
print(f"  R.from_euler('XYZ', np.deg2rad({np.round(eul, 2).tolist()}))")
````

### 4.4 叠图
仿真边缘画成**红线**,青色框为策略可见区域。

````python
def overlay(real_path, sim_path, crop_box=True):
    real, sim = cv2.imread(str(real_path)), cv2.imread(str(sim_path))
    edges = cv2.Canny(cv2.cvtColor(sim, cv2.COLOR_BGR2GRAY), 50, 150)
    out = cv2.addWeighted(real, 0.6, sim, 0.4, 0); out[edges > 0] = (0, 0, 255)
    if crop_box: cv2.rectangle(out, (96, 16), (544, 464), (255, 255, 0), 1)
    return cv2.cvtColor(out, cv2.COLOR_BGR2RGB)

fig, ax = plt.subplots(1, 2, figsize=(16, 6))
ax[0].imshow(overlay(ALIGN/"real_front.png", ALIGN/"sim_front.png")); ax[0].set_title("FRONT overlay")
ax[1].imshow(overlay(ALIGN/"real_wrist.png", ALIGN/"sim_wrist.png", crop_box=False)); ax[1].set_title("WRIST overlay")
for a in ax: a.axis("off")
plt.show()
cv2.imwrite(str(ALIGN/"overlay_front.png"), cv2.cvtColor(overlay(ALIGN/"real_front.png", ALIGN/"sim_front.png"), cv2.COLOR_RGB2BGR))
````

### 4.5 调整外参(在 notebook 里完成,循环做)

**达标标准**:青框内,桌面边缘、挡条、零件、手指的红线和真实边缘偏差 ≤ **约 5 像素**(0.7 m 处 1° ≈ 10 px)。

**一轮的流程**
1. 运行 **A 格**:读出渲染脚本里当前的前视/腕部外参,换算成基座系位置和俯角、偏航,方便对照。
2. 看叠图,决定怎么改,在 **B 格**里改数字(基座系、米和度),运行。它会备份并改写渲染脚本,重新打包 `to_isaac.tar.gz`。
3. 拷到 Isaac 电脑渲染(4.3 的命令块,注意解压后要重做 `import numba` 那一行)→ 拷回 `align_out` → 重新运行取帧格和叠图格。

**怎么判断往哪边调**(相机朝机器人,画面上方是远处):
| 看到的现象 | 调什么 |
|---|---|
| 真实内容比仿真**低**(靠画面下方) | **减小**俯角 `PITCH`(相机更平视,仿真内容会往下移) |
| 真实内容比仿真**高** | 增大 `PITCH` |
| 真实内容比仿真偏**左/右** | 调 `YAW`,方向没有现成结论,先试 ±1°,看红线往哪边动,再决定正负 |
| 真实内容比仿真**大**(更近) | 把相机往后挪:增大 `POS_X` |
| 整体尺寸对但上下错位 | 先调 `PITCH`,再用 `POS_Z` 微调 |

调的顺序:**先俯仰和偏航,让整体位置对上;再平移,让大小和透视对上;最后滚转**(本 notebook 不含滚转,通常不需要)。每次只改一个量,步长:角度 1–3°,位置 2–5 cm。

````python
import re
RR = REPO / "src" / "sim2real" / "isaac_lab_rerender.py"
BASE_TO_WORLD = np.array([-0.30, 0.0, 0.415]); FLIP = np.diag([1, -1, -1])

def _front_block(txt):
    m = re.search(r"cam_pos = torch\.from_numpy\(np\.array\(\[([^\]]*)\]\)\.reshape\(1, 3\)\).*?camera\._view\.set_world_poses", txt, re.S)
    return m
def read_front(txt):
    blk = _front_block(txt).group(0)
    pos = [float(x) for x in re.search(r"np\.array\(\[([^\]]*)\]\)", blk).group(1).split(",")]
    eul = [float(x) for x in re.search(r"np\.deg2rad\(\[([^\]]*)\]\)", blk).group(1).split(",")]
    return np.array(pos), np.array(eul)

def pose_from_render(pos_w, eul):
    """渲染器格式 → 基座系位置、俯角、偏航、滚转(度)"""
    Rcv = R.from_euler("XYZ", eul, degrees=True).as_matrix() @ FLIP
    f = Rcv[:, 2]
    pitch = np.degrees(np.arcsin(-f[2])); yaw = np.degrees(np.arctan2(-f[1], -f[0]))
    ref = look_pose([0, 0, 0], pitch, yaw)[:3, :3]
    roll = R.from_matrix(ref.T @ Rcv).magnitude() * 180 / np.pi   # 与无侧倾姿态的夹角
    return pos_w - BASE_TO_WORLD, pitch, yaw, roll

def look_pose(pos, pitch_deg, yaw_deg=0.0):
    th, ps = np.radians(pitch_deg), np.radians(yaw_deg)
    z = np.array([-np.cos(th)*np.cos(ps), -np.cos(th)*np.sin(ps), -np.sin(th)])
    x = np.cross(z, [0, 0, 1]); x /= np.linalg.norm(x); y = np.cross(z, x)
    T = np.eye(4); T[:3, :3] = np.c_[x, y, z]; T[:3, 3] = pos; return T

pw, ew = read_front(RR.read_text())
pb, pitch0, yaw0, roll0 = pose_from_render(pw, ew)
print(f"[A] 渲染脚本里的前视:  cam_pos(世界)={pw.tolist()}  euler XYZ={ew.tolist()}")
print(f"    换算到基座系:  POS_BASE=({pb[0]:.3f}, {pb[1]:.3f}, {pb[2]:.3f}) m   PITCH={pitch0:.1f}°  YAW={yaw0:.1f}°  (残余滚转 {roll0:.2f}°,应≈0)")
````

````python
# ===== B 格:改这里(基座系,米/度)。默认等于当前脚本里的值 =====
FRONT_POS_BASE = [round(float(v), 3) for v in pb]
FRONT_PITCH    = 28 # round(float(pitch0), 1)
FRONT_YAW      = round(float(yaw0), 1)
# 例:真实内容偏低 → FRONT_PITCH = 30.0;真实内容偏大 → FRONT_POS_BASE[0] += 0.05
# ================================================================

import shutil, tarfile
def apply_front(pos_base, pitch, yaw):
    T = look_pose(pos_base, pitch, yaw)
    pos_w = T[:3, 3] + BASE_TO_WORLD
    eul = R.from_matrix(T[:3, :3] @ FLIP).as_euler("XYZ", degrees=True)
    txt = RR.read_text(); m = _front_block(txt); blk = m.group(0)
    new = re.sub(r"np\.array\(\[[^\]]*\]\)\.reshape\(1, 3\)", f"np.array({np.round(pos_w, 4).tolist()}).reshape(1, 3)", blk, count=1)
    new = re.sub(r"np\.deg2rad\(\[[^\]]*\]\)", f"np.deg2rad({np.round(eul, 3).tolist()})", new, count=1)
    bak = RR.with_suffix(".py.bak")
    if not bak.exists(): shutil.copy(RR, bak)            # 只在第一次备份原文件
    RR.write_text(txt.replace(blk, new))
    return pos_w, eul

pos_w, eul = apply_front(FRONT_POS_BASE, FRONT_PITCH, FRONT_YAW)
print("已写入渲染脚本:  cam_pos =", np.round(pos_w, 4).tolist(), "  euler XYZ =", np.round(eul, 3).tolist())
print("(原文件备份在", RR.with_suffix(".py.bak").name, ")")

bundle = REPO / "scratch" / "to_isaac.tar.gz"
with tarfile.open(bundle, "w:gz") as t:
    t.add(RR, arcname="src/sim2real/isaac_lab_rerender.py"); t.add(ALIGN/"align.pkl", arcname="scratch/align/align.pkl")
print("已重新打包:", bundle, " → 拷到 Isaac 电脑,按 4.3 的命令块渲染")
````
